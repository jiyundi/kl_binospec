```python
paths = {
    'A1': Path('path/to/A1/spec2d.fits'),
    'A2': Path('path/to/A2/spec2d.fits'),
    'A3': Path('path/to/A3/spec2d.fits'),
    'B':  Path('path/to/B/spec2d.fits'),
    'C':  Path('path/to/C/spec2d.fits'),
}
line_name = 'O2'
line_waves = [3727.092, 3729.875]  # Angstrom
redshift = 1.0  # replace with secure redshift from the table
spec_width, spec_height = 40, 20
plan_path = Path('slit_007_crops.json')
```

```python
sources = load_sources(paths)
review = NotebookCropReview(sources, line_name, line_waves, redshift,
                            spec_width, spec_height, plan_path)
review.show()
```
